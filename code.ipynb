```python
import gensim
```

```python
from gensim.models import KeyedVectors
```

```python
path="D:\GoogleNews-vectors-negative300.bin"
model=KeyedVectors.load_word2vec_format(path,binary=True,limit=200000)
```

```python
model['man']
```

```python
model['king']-model['queen']
```

```python
model.similarity('man','child')
```

```python
model.doesnt_match(['PHP','Java','monkey'])
```

```python
model.doesnt_match(['cricket','snake','football','hockey','bottle','swimming','pen','paper','lunchbox','diary'])
```

```python
model.most_similar('king')
```

```python
vec=model['king'] - model['man'] + model['woman']
model.most_similar([vec])
```

```python
vec=model['INR'] - model['India'] + model['USA']
model.most_similar([vec])
```

```python
vec=model['football']+model['hockey']
model.most_similar([vec])
```
